## 1. Yield


```scala
//Used in for-comprehensions to produce a collection
val numbers = List(1,2,3,4,5,6)
val doubled = for (n <- numbers) yield n * 2
println(doubled) 
```

## 2. Predicate

```scala
//a function that returns Boolean
val numbers = List(1, 2, 3, 4, 5)
val evenNumbers = numbers.filter(_ % 2 == 0)
println(evenNumbers)
```

## 3. Partial Applications

```scala
//Fix some arguments of a function to create a new function
def add(a: Int, b: Int) = a + b
val addFive = add(5, _: Int)
println(addFive(13))

def sendMessage(tag: String, message: String) = s"$tag: $message"
val alertMessage = sendMessage("ALERT", _: String)
println(alertMessage("Door is open"))
```

## 4. Function Composition

```scala
val f = (x: Int) => x + 1
val g = (x: Int) => x * 2

val h1 = f andThen g  // g(f(x))
println(h1(5))  

val h2 = f compose g  // f(g(x))
println(h2(5))

//Note:
    //andThen ---> Left-to-right execution-- output of first function is input to the second
    //compose ---> Right-to-left execution--- output of second function is input to the first
```

## 5. Blocks

```scala
//Group of expressions inside {}. last expression is returned
val result = {
  val a = 10
  val b = 20
  a + b
}
println(result)

//Note:
    //Blocking--->Waits for a result before continuing
    //Non-Blocking---->Continues execution while waiting.
```

## 6. Exception Handling

```scala
try {
  val x = 10 / 0
} catch {
  case e: ArithmeticException => "Cannot divide by zero"
} finally {
  println("Execution finished")
}

//No throws keyword.
//Use try,catch, finally
```

## 7. Future

```scala
import scala.concurrent._
import ExecutionContext.Implicits.global
import scala.util.{Success, Failure}


val squareFuture = Future {
  5 * 5
}


squareFuture.onComplete {
  case Success(value) => println(s"Square is: $value")
  case Failure(e) => println(s"Error: ${e.getMessage}")
}
```

## 8. Implicit

```scala

implicit val multiplier: Int = 3
def multiply(x: Int)(implicit m: Int): Int = x * m

//call without passing the second argument
val result = multiply(5)
println(result) 
```
