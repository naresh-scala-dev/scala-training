## 1. Lazy Evaluation

```scala
// value will not be calculated until it is used
lazy val city = {
  println("Loading city name...")
  "Chennai"
}
println("before using city")
println(city)      // evaluated here
println(city)      // not printed again since already evaluated
```

## 2. Tail Recursion

```scala
// function calling itself at the end (optimized by compiler)
def factorial(n: Int, acc: Int = 1): Int = {
  if (n <= 1) acc
  else factorial(n - 1, n * acc)
}
println(factorial(5))  // 120
```

## 3. Function Returning Function

```scala
def greet(name: String): () => String = {
  () => s"Hello $name"
}
val helloFunc = greet("Chennai")
println(helloFunc())  // calling inner function
```

## 4. Function with Keyword Parameters

```scala
def displayInfo(name: String, age: Int): Unit = {
  println(s"My name is $name and I am $age years old")
}
displayInfo(age = 25, name = "Naresh")
```

## 5. Function with Variable Number of Parameters

```scala
def printFruits(fruits: String*): Unit = {
  for (fruit <- fruits)
    println(fruit)
}
printFruits("Apple", "Mango", "Banana")
```

## 6. Higher Order Functions (HOF)

```scala
// function takes another function as input or returns one
def processNumbers(nums: List[Int], f: Int => Int): List[Int] = {
  nums.map(f)
}
val numbers = List(1, 2, 3, 4)
println(processNumbers(numbers, x => x * 2))
```

## 7. Common Collection Methods

```scala
val nums = List(1, 2, 3, 4, 5, 6)

// map ----->apply a function to every element
val doubled = nums.map(_ * 2)
println(doubled)   // List(2, 4, 6, 8, 10, 12)

val squared = nums.map(n => n * n)
println(squared)   // List(1, 4, 9, 16, 25, 36)


// filter----> keep only elements that satisfy a condition
val evens = nums.filter(_ % 2 == 0)
println(evens)     // List(2, 4, 6)

val greaterThanThree = nums.filter(_ > 3)
println(greaterThanThree)   // List(4, 5, 6)


// reduce ---> combine all elements into one
val sum = nums.reduce(_ + _)
println(sum)       // 21

val product = nums.reduce(_ * _)
println(product)   // 720


// foldLeft ----> accumulate from left with a starting value
val addFrom10 = nums.foldLeft(10)(_ + _)
println(addFrom10)  // 31

val stringJoin = nums.foldLeft("Start:")((acc, n) => s"$acc $n")
println(stringJoin)  // Start: 1 2 3 4 5 6


// foldRight ----> same as foldLeft but starts from right
val foldRightExample = nums.foldRight(0)(_ - _)
println(foldRightExample)  // (((((1-2)-3)-4)-5)-6) = -19

val rightString = nums.foldRight("End")((n, acc) => s"$n -> $acc")
println(rightString)  // 1 -> 2 -> 3 -> 4 -> 5 -> 6 -> End


// scanLeft ---> keeps running total (shows all steps)
val runningSum = nums.scanLeft(0)(_ + _)
println(runningSum)  // List(0, 1, 3, 6, 10, 15, 21)

val runningProduct = nums.scanLeft(1)(_ * _)
println(runningProduct)  // List(1, 1, 2, 6, 24, 120, 720)

// scanRight ----> same as scanLeft but from right
val scanRightSum = nums.scanRight(0)(_ + _)
println(scanRightSum)  // List(21, 20, 18, 15, 11, 6, 0)

val scanRightProduct = nums.scanRight(1)(_ * _)
println(scanRightProduct)  // List(720, 720, 360, 120, 30, 6, 1)


// collect ----> used for partial filtering and transformation
val evenTimes10 = nums.collect { case x if x % 2 == 0 => x * 10 }
println(evenTimes10) // List(20, 40, 60)

val oddAsText = nums.collect {
  case x if x % 2 != 0 => s"Odd number: $x"
}
println(oddAsText) // List(Odd number: 1, Odd number: 3, Odd number: 5)
```

## 8. Partial Application

```scala
//Partial Functions
//A partial function works only for some inputs, not for all

val divide: PartialFunction[Int, Int] = {
  case x if x != 0 => 100 / x
}

println(divide.isDefinedAt(10))  // true
println(divide.isDefinedAt(0))   // false

println(divide(5))   // 20

//Safe usage
if (divide.isDefinedAt(0))
  println(divide(0))
else
  println("Cannot divide by zero")


val dayType: PartialFunction[String, String] = {
  case "Saturday" | "Sunday" => "Weekend"
  case "Monday" | "Tuesday" | "Wednesday" | "Thursday" | "Friday" => "Weekday"
}

println(dayType("Sunday"))
println(dayType("Tuesday"))


//Combine partial functions using orElse
val numbers: PartialFunction[Int, String] = {
  case x if x > 0 => "Positive"
}
val negatives: PartialFunction[Int, String] = {
  case x if x < 0 => "Negative"
}

val result = numbers orElse negatives
println(result(5))
println(result(-3))
```

## 9. Currying in Functions

```scala
//Currying Functions
//A curried function breaks parameters into multiple groups

def add(a: Int)(b: Int): Int = a + b

println(add(5)(10))  // both params together

//Partial application —--> fix one parameter and use later
val add5 = add(5)    // now add5 is a function waiting for 2nd param
println(add5(20))    // 25
println(add5(50))    // 55


def greet(name: String)(city: String): String =
  s"Hello $name, welcome to $city!"

println(greet("Nithin")("Chennai"))

//Partial usage
val greetPerson = greet("Ravi")    // city is not yet given
println(greetPerson("Coimbatore"))
println(greetPerson("Delhi"))


def totalPrice(price: Double)(taxRate: Double): Double =
  price + (price * taxRate / 100)

println(totalPrice(100)(10))  // 110.0

//Partial function — fix taxRate for reuse
val gst = totalPrice(_: Double)(18)  // use underscore for placeholder
println(gst(1000))   // 1180.0
println(gst(250))    // 295.0
```
