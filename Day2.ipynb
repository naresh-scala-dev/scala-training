## 1. Classes and Objects

```scala
// Classes and Objects: blueprint for creating objects

// 1. Normal class with primary constructor
class City(var name: String, var population: Int) {
  def info(): String = s"City: $name, Population: $population"
  def updatePopulation(newPop: Int): Unit = { population = newPop }
}

// 2. Class with default values in constructor
class Town(var name: String = "Unknown", var population: Int = 0) {
  def info(): String = s"Town: $name, Population: $population"
}

// 3. Class extending another class
class MetroCity(name: String, population: Int, var metroArea: String)
  extends City(name, population) {

  // Overriding method
  override def info(): String = s"MetroCity: $name, Population: $population, Area: $metroArea"
}

// Creating objects
val chennai = new City("Chennai", 10000000)
val mumbai = new City("Mumbai", 20000000)
val pondy = new Town()          // uses default values
val delhi = new MetroCity("Delhi", 15000000, "NCR")

// Accessing methods
println(chennai.info())
println(mumbai.info())
println(pondy.info())
println(delhi.info())

// Updating data
chennai.updatePopulation(11000000)
println(chennai.info())

// Companion object example
object City {
  def greeting(cityName: String): String = s"Welcome to $cityName!"
}

println(City.greeting("Chennai"))

// Using @main function style (Scala 3)
@main def runCities(): Unit = {
  println("Running Cities Program")
  println(chennai.info())
  println(delhi.info())
}
```
