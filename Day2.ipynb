## 2. Normal class vs Case Class


```scala
//case classes provide automatic equals, hashCode, toString, copy and pattern matching.

class NormalPerson(val name: String, val age: Int)
case class CasePerson(name: String, age: Int)

val p1 = new NormalPerson("Arun", 25)
val p2 = new NormalPerson("Arun", 25)
println(p1 == p2)   // false

val c1 = CasePerson("Arun", 25)
val c2 = CasePerson("Arun", 25)
println(c1 == c2)   // true

val c3 = c1.copy(age = 30)
println(c3)
```

## 3. apply and unapply methods

```scala
//apply creates objects without `new`, unapply extracts values for pattern matching.

case class Employee(name: String, salary: Int)

val e1 = Employee("Naresh", 50000) // apply
println(e1.name + " - " + e1.salary)

e1 match {
  case Employee(n, s) => println(s"Name: $n, Salary: $s")  // unapply
}

val e2 = Employee("Priya", 60000)
e2 match {
  case Employee(n, s) => println(s"$n earns $s")
}
```

## 4. Companion Object

```scala
//Companion objects share the same name as the class and can access its private members.

class Student(val name: String) {
  def greet(): Unit = println(s"Hello, $name!")
}

object Student {
  def apply(name: String): Student = new Student(name)
  def course(): Unit = println("Scala Training in Progress")
}

val st1 = Student("Mani")
st1.greet()
Student.course()

val st2 = Student("Anu")
st2.greet()
Student.course()
```

## 5. Operator Overloading

```scala
Y//ou can redefine operators as methods to work with your custom classes.

class Counter(val count: Int) {
  def +(other: Counter): Counter = new Counter(this.count + other.count)
  def -(other: Counter): Counter = new Counter(this.count - other.count)
}

val c1 = new Counter(10)
val c2 = new Counter(5)
val sumCounter = c1 + c2
val diffCounter = c1 - c2

println(sumCounter.count)  // 15
println(diffCounter.count) // 5

val c3 = new Counter(20)
val c4 = new Counter(8)
println((c3 + c4).count)  // 28
println((c3 - c4).count)  // 12
```

## 6. Methods with Default Parameters

```scala
//Default values make parameters optional when calling methods.

def greetPerson(name: String = "Guest", city: String = "Chennai"): Unit =
  println(s"Hello $name from $city!")

greetPerson()
greetPerson("Naresh")
greetPerson("Priya", "Bangalore")

greetPerson(city = "Mumbai")
greetPerson("Ravi", city = "Kolkata")
```

## 7. Method Overloading

```scala
//multiple methods can have the same name but different parameter lists.

class Calculator {
  def add(x: Int, y: Int): Int = x + y
  def add(x: Double, y: Double): Double = x + y
  def add(x: Int, y: Int, z: Int): Int = x + y + z
}

val calc = new Calculator()
println(calc.add(5, 10))
println(calc.add(2.5, 3.5))
println(calc.add(1, 2, 3))

val calc2 = new Calculator()
println(calc2.add(100, 200))
println(calc2.add(10.5, 20.5))
println(calc2.add(3, 4, 5))
```
