## 3. Conditional Statements

```scala
//CONDITIONAL STATEMENTS
//use to execute the logic based on condition

//if
val num = 36

if (num > 35)
  println("Pass")

//if else

val x = -5

if (x > 0)
  println("Positive number")
else
  println("Negative number")

//else-if


//curly braces are optional for single line expressions but recommened for multiple lines
val marks = 99

val grade =
  if (marks >= 90) {
    println("Excellent")
    "A"         //-------- no need of explicit return
  } else if (marks >= 80) {
    println("Very Good")
    "B"
  } else if (marks >= 70) {
    println("Good")
    "C"
  } else {
    println("Needs Improvement")
    "Fail"
  }
println(s"Grade: $grade")

//NOTE:--No return needed — scala automatically returns the value of the block.
```

```scala
//Boolean Expressions
// && ----> logical AND
// || ----> logical OR
// !  ----> ogical NOT

val a = true
val b = false
val c = true

if (a && b) {
  println("Both a and b are true")
} else if (a || b) {
  println("At least one of a or b is true")
} else if (!c) {
  println("c is false")
} else {
  println("None of the above")
}
```

```scala
//pattern matching(Like switch case)

val x = 2

val result = x match {
  case 1 => "One"
  case 2 => "Two"
  case 3 => "Three"
  case _ => "Other"
}

println(result)


val marks = 82
val grade = marks match
  case x if x >= 90 => "A"
  case y if y >= 80 => "B"
  case z if z >= 70 => "C"
  case _ => "Fail"   //----"_" denotes defult case like if none of the case matches it enter into this case

println(grade)

//note: if Add extra guard condition
```

## 4. Loops(Control Structures)

```scala
//loops---->perfom the task multiple times unti a condition becomes false

//for loop

for (i <- 1 to 5) //<- means  <= (1,2,3,4,5)
  println(s"Value of i = $i") 

for (i <- 1 until 5) //until is < (1,2,3,4)
  println(i)

for (i <- 1 to 10 by 2) //by for step size--> increased by 2 
  println(i)

//foreach
val fruits = List("Apple", "Banana", "Cherry")

for (fruit <- fruits)
  println(fruit)

//nested loop
for (i <- 1 to 3; j <- 1 to 2) //to --->includes end value
  println(s"i = $i, j = $j")

//Loop Guards (Filters)
for (num <- 1 to 10 if num % 2 == 0)
  print(s"even numbers---$num ")

println()

//while loop
var i = 0
while (i < 5) {
  println(s"i = $i")
  i += 1
}
```

## 5. Arrays

```scala
// An Array is a fixed-size collection that stores elements of the same type

// Create an array with values
val numbers = Array(10, 20, 30, 40, 50)
val names = Array("Scala", "Java", "Python")

// update elements based on index
numbers(1) = 25
println(numbers(1))   // prints 25

// Get length of array using length
println(numbers.length)   // prints 5

// Loop through arrays using for
for (n <- numbers)
  println(n)

// Loop using index
for (i <- numbers.indices)
  println(s"Index $i ----> ${numbers(i)}")

// Using foreach 
numbers.foreach(n => println(n))
```

## 6. Collections

```scala
//Collections are used to store, manipulate and provide multiple values together(List,Set,Queue)

// Lists (Immutable)
// Create a list with initial values
val fruits = List("Apple", "Banana", "Cherry")

// Access elements
println(fruits(0))  

// Add elements (creates a new list)
val newFruits = "Orange" :: fruits
println(newFruits)  

// Loop through the list
for (fruit <- fruits)
  println(fruit)


// ArrayBuffer (Mutable List)
import scala.collection.mutable.ArrayBuffer

// Create an ArrayBuffer with initial values
val numbers = ArrayBuffer(10, 20, 30)

// Add element
numbers += 40

// Remove element
numbers -= 20

// Update element
numbers(0) = 100

// Print the ArrayBuffer
println(numbers.mkString(", "))  // 100, 30, 40


// Set (Unique elements)
// Immutable set
val set1 = Set(1, 2, 3, 2, 1)
println(set1)  // 1, 2, 3

// Mutable set
import scala.collection.mutable.Set
val set2 = Set(1, 2, 3)
set2 += 4
set2 -= 2
println(set2.mkString(", "))  // 1, 3, 4


// Map (Key-Value pairs)
// Immutable map
val capital = Map("India" -> "New Delhi", "USA" -> "Washington")
println(capital("India"))  // New Delhi

// Mutable map
import scala.collection.mutable.Map
val scores = Map("Alice" -> 90, "Bob" -> 85)

// Add new key-value
scores("Charlie") = 95

// Update value
scores("Bob") = 88

// Print map
for ((k, v) <- scores)
  println(s"$k scored $v")
```

## 7. Functions

```scala
// Functions: reusable code blocks

// Basic function
def greetCity(): String = "Welcome to Chennai!"
println(greetCity())

// Function with default parameter
def welcome(place: String = "Chennai"): String = s"Hello from $place!"
println(welcome())
println(welcome("Hyd"))

// Function overloading
def showInfo(city: String): String = s"City: $city"
def showInfo(city: String, population: Int): String = s"City: $city, Population: $population"
println(showInfo("Chennai"))
println(showInfo("Chennai", 10000000))

// Function returning Unit (no value)
def printWelcome(city: String): Unit = println(s"Visiting $city is fun!")
printWelcome("Chennai")
```
